```python
#==============================================================================================
# Simulation Study: Row-after-Column CUR (p = 1000, n = 1000, reps = 10)
#==============================================================================================

data_folder = "Simulation Data/p1000_n1000"
results_folder = "Results/column_row/p1000_n1000"

print("Running Row-after-Column CUR sampling-variance study...")

results_1000_cr = run_sampling_variance_row_after_col(
    k_vector=k_vector,   
    base=base,
    folder=data_folder,
    reps=10,             
    outer_reps=10,       
    test_size=0.2,       
    row_reduction=True,
    results_folder = results_folder,
    save_name="results_cr_p1000_n1000"
)

save_results(
    f"{base}/{results_folder}/results_cr_p1000_n1000.pkl",
    results=results_1000_cr
)

print("Saved Row-after-Column CUR results.")
```
